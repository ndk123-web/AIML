```python
import pandas as pd 
```

```python
# Datasets
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
import matplotlib as plt
import seaborn as sn
```

```python
digits = load_digits()
```

```python
dir(digits)
```

```python
len(digits)
```

```python
len(digits.data)
```

```python
len(digits.images)
digits.target
```

```python
 X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.3, random_state=42)
```

```python
len(X_train)
```

```python
model = LogisticRegression()
```

```python
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
model.score(X_test, y_test)
```

```python
model.coef_
```

```python
model.intercept_
```

# where our model is lacking

- to see that we can use confusion matrix

```python
y_predicted = model.predict(X_test)

cm = confusion_matrix(y_test, y_predicted)
```

# What this Confusion Matrix telling us ?

example: 
- see at location (0,0) = 53 it is telling that actual is 0 and predicted is 0 such output detected are 53 (means its good)
- now see at (1,2) = 1 it means actual is 1 but predicted is 2 that kind of output detected 1 time (means its error or false class)
