```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("homeprices.csv")

sx = MinMaxScaler()
sy = MinMaxScaler()
```

```python
X = df.drop("price" , axis = 1)
y = df["price"]
```

```python
scaled_x = sx.fit_transform(X)
scaled_y = sy.fit_transform(y.values.reshape(df.shape[0],1))

scaled_x
```

```python
scaled_x
scaled_y

new_X = pd.DataFrame(scaled_x, columns=["area", "bedrooms", "age"])
scaled_y = scaled_y.flatten()
```

## Batch Gradient
- 1 Epoch 1 Total Error mean caluclted
- at 1 epoch 1 time weight is going to be updated

```python
def batch_gradient_descent(X, y, epochs):

    w1 = w2 = w3 = 1
    bias = 1
    learning_rate = 0.1
    n = len(X)

    for i in range(epochs):

        y_predicted = (
            w1 * X["age"] +
            w2 * X["bedrooms"] +
            w3 * X["area"] +
            bias
        )

        cost = (1/n) * sum((y - y_predicted) ** 2)

        w1d = -(2/n) * sum(X["age"] * (y - y_predicted))
        w2d = -(2/n) * sum(X["bedrooms"] * (y - y_predicted))
        w3d = -(2/n) * sum(X["area"] * (y - y_predicted))
        biasd = -(2/n) * sum(y - y_predicted)

        w1 = w1 - learning_rate * w1d
        w2 = w2 - learning_rate * w2d
        w3 = w3 - learning_rate * w3d
        bias = bias - learning_rate * biasd

        print(f"Epoch:{i}, Loss:{cost}")

    return w1, w2, w3, bias
```

```python
batch_gradient_descent(new_X,scaled_y,50)
```

## Stochastic Gradient Descent
- For 1 epoch its going to updated weights for each N sample (N is size of X and y)

```python
def stochastic_gradient_descent(X, y, epochs):

    w1 = w2 = w3 = 1
    bias = 1
    learning_rate = 0.01
    n = len(X)

    for epoch in range(epochs):

        for i in range(n):

            x1 = X["age"].iloc[i]
            x2 = X["bedrooms"].iloc[i]
            x3 = X["area"].iloc[i]
            yi = y[i]

            y_pred = w1*x1 + w2*x2 + w3*x3 + bias

            error = yi - y_pred

            w1d = -2 * x1 * error
            w2d = -2 * x2 * error
            w3d = -2 * x3 * error
            biasd = -2 * error

            w1 -= learning_rate * w1d
            w2 -= learning_rate * w2d
            w3 -= learning_rate * w3d
            bias -= learning_rate * biasd

        print(f"Epoch {epoch}")

    return w1, w2, w3, bias
```

```python
stochastic_gradient_descent(new_X, scaled_y, 50)
```

## Mini Batch Gradient Descent
- For 1 epochs it groups N batches (N single Samples) and then update the weights 

```python
def mini_batch_gradient_descent(X, y, epochs, batch_size):

    w1 = w2 = w3 = 1
    bias = 1
    learning_rate = 0.01
    n = len(X)

    for epoch in range(epochs):

        count = 1
        for start in range(0, n, batch_size):

            end = start + batch_size

            X_batch = X.iloc[start:end]
            y_batch = y[start:end]

            y_pred = (
                w1 * X_batch["age"] +
                w2 * X_batch["bedrooms"] +
                w3 * X_batch["area"] +
                bias
            )

            error = y_batch - y_pred

            w1d = -(2/len(X_batch)) * sum(X_batch["age"] * error)
            w2d = -(2/len(X_batch)) * sum(X_batch["bedrooms"] * error)
            w3d = -(2/len(X_batch)) * sum(X_batch["area"] * error)
            biasd = -(2/len(X_batch)) * sum(error)

            w1 -= learning_rate * w1d
            w2 -= learning_rate * w2d
            w3 -= learning_rate * w3d
            bias -= learning_rate * biasd
            print(f"Epoch {epoch}, Batch: {count} , Cost: {error}")
            count+=1
        count = 1

    return w1, w2, w3, bias
```

```python
mini_batch_gradient_descent(new_X, scaled_y, 50, 2)
```
