```python
from pydantic import BaseModel, Field, Annotated, operator
```

```python
class Task(BaseModel):
    id: int
    title: str
    brief: Field(...,description="what to cover")
```

```python
class Plan(BaseModel):
    blog_title: str
    tasks: List[Task]
```

```python
class State(BaseModel):
    topic: str
    plan: Plan
    sections: Annonated[List[str], operator.add]
    final: str
```
