```python
from gensim.models import Word2Vec
```

```python
# Tokenized dataset (list of lists)
sentences = [
    ["the", "king", "rules", "the", "kingdom"],
    ["the", "queen", "rules", "the", "empire"],
    ["a", "man", "walks", "into", "a", "bar"],
    ["a", "woman", "walks", "into", "a", "restaurant"]
]
```

```python
# sg=1 means Skip-Gram
model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, sg=1)
```

```python
print("Embedding for the 'King': ", model.wv["king"])
```

```python
# Find words with the most similar context vectors
print("Most similar to 'king':", model.wv.most_similar('king', topn=2)) 
```

```python
# Perform vector arithmetic: King - Man + Woman
result = model.wv.most_similar(positive=['king', 'woman'], negative=['man'], topn=1)
print("Arithmetic Result:", result)
```
