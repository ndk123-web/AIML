```python
import torch 
```

```python
x = torch.tensor(3.0, requires_grad=True)
y = torch.tensor(4.0, requires_grad=True) 
```

```python
a = x ** 2      # a = 3^2 = 9
a.retain_grad()   # don't free leaf tensors , let it be there for deubgging purpose 

b = 2 * a           # b = 2 * 9 = 18
b.retain_grad() 

c = 3 * y           # c = 3 * 4 = 12
c.retain_grad() 

z = b + c + 5       # z = 18 + 12 + 5 = 35
z.retain_grad()

print(f"Forward Output z: {z.item()}")
```

```python
print("How 'z' was made:", z.grad_fn)  # <AddBackward0> (Addition)
print("How 'b' was made:", b.grad_fn)  # <MulBackward0> (Multiplication)
print("How 'a' was made:", a.grad_fn)  # <PowBackward0> (Power)
print("How 'x' was made:", x.grad_fn)  # None (Leaf tensor created by user)
```

```python
# Before backward pass
print("x.grad before backward:", x.grad)  # None
print("y.grad before backward:", y.grad)  # None

# Trigger backpropagation
z.backward(retain_graph=True)


'''
    After z.backward() , triggers backpropagation,

    z = b + c + 5

    dZ/dx = dZ/db * db/da * da/dx 
    -> 1 * 2 * 2*x 
    -> 1 * 2 * 2*3
    -> 12

    dZ/dy = dZ/dc * dc/dy
    -> 1 * 3
    -> 3  
'''

# After backward pass
print(f"Computed dz/dx: {x.grad}")  # Expected: 4 * 3.0 = 12.0
print(f"Computed dz/dy: {y.grad}")  # Expected: 3.0
print(f"Computed dz/da: {a.grad}")  # Expected: 2.0
print(f"Computed dz/db: {b.grad}")  # Expected: 1.0
print(f"Computed dz/dc: {c.grad}")  # Expected: 1.0 
```
