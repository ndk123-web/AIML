```python
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

raw_text = "Natural Language Processing with NLTK is amazing! The dogs were running quickly across fields."

# 1. Sentence and Word Tokenization
sentences = sent_tokenize(raw_text)
tokens = word_tokenize(raw_text.lower()) # Lowercasing + splitting

# 2. Stop Words & Punctuation Removal
stop_words = set(stopwords.words('english'))
cleaned_tokens = [
    w for w in tokens 
    if w not in stop_words and w not in string.punctuation
]

# 3. Stemming vs Lemmatization
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stemmed = [stemmer.stem(w) for w in cleaned_tokens]
lemmatized = [lemmatizer.lemmatize(w, pos='v') for w in cleaned_tokens]

print("Tokens:", cleaned_tokens)
print("Stemmed:", stemmed)       # Chop off suffixes crudely: 'running' -> 'run'
print("Lemmatized:", lemmatized) # Dictionary-based root form: 'running' -> 'run'
```

```python
from nltk.corpus import brown

# 1. View available categories/genres
print("Categories:", brown.categories()[:5]) 
# ['adventure', 'belles_lettres', 'editorial', 'fiction', 'government']

# 2. Extract words and sentences for a specific genre
news_words = brown.words(categories='news')
print("Total words in news:", len(news_words))
print("First 10 words:", news_words[:10])

# 3. Tagged sentences (Words with their Ground-Truth POS tags)
tagged_sents = brown.tagged_sents(categories='news')
print("\nSample Tagged Sentence:")
print(tagged_sents[0][:5]) # [('The', 'AT'), ('Fulton', 'NP-TL'), ('County', 'NN-TL'), ...]
```

```python
import nltk

sentence = "The small dog barked at the mailman"
tokens = nltk.word_tokenize(sentence)

# 1. Generate POS Tags
pos_tags = nltk.pos_tag(tokens)
print("POS Tags:\n", pos_tags)

# 2. Define a Chunk Rule (Noun Phrase Chunking)
# NP: {<DT>?<JJ>*<NN>} -> Optional Determiner, 0+ Adjectives, ending with Noun
chunk_grammar = r"NP: {<DT>?<JJ>*<NN>}"
chunk_parser = nltk.RegexpParser(chunk_grammar)

chunk_tree = chunk_parser.parse(pos_tags)
print("\nChunk Tree Structure:")
print(chunk_tree)
```

```python
from nltk import CFG

# 1. Define explicit CFG production rules
toy_grammar = CFG.fromstring("""
  S -> NP VP
  NP -> DT JJ NN | DT NN | PRP
  VP -> VBD NP | VBD PP | VBD
  PP -> IN NP
  DT -> 'the' | 'The' | 'a'
  JJ -> 'small'
  NN -> 'dog' | 'mailman'
  VBD -> 'barked' | 'chased'
  IN -> 'at'
  PRP -> 'he'
""")

# 2. Parse a valid input sentence
sentence_tokens = "The small dog chased the mailman".split()
rd_parser = nltk.RecursiveDescentParser(toy_grammar)

print("\n--- CFG Parse Trees ---")
for tree in rd_parser.parse(sentence_tokens):
    print(tree)
    # tree.draw() # (Opens a UI visualization window of the syntax tree)
```
