```python
import matplotlib.pyplot as plt
from IPython.display import clear_output
import time
import matplotlib.animation as animation
```

```python
%matplotlib inline
```

```python
x = [1,2,3,4,5,6,7,8]
y = [1,4,9,16,25,36,49,64]

for i in range(50):
    for j in range(8):
        x[j] = x[j] + 1
        y[j] = y[j] - 1
    
    clear_output(wait=True)      # clear previous frame
    plt.plot(x, y, marker='o', linewidth=5, linestyle="dotted")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(f"Iteration {i}")
    plt.ylim(0, 70)
    plt.show()
    
    time.sleep(0.1)
        
```
