```python
import numpy as np
import math 

# always between 0 and 1, so it is not a problem for backpropagation
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
```

```python
sigmoid(100) # it is close to one but not exactly one, so it is not a problem for backpropagation
sigmoid(0) # it is exactly 0.5, so it is not a problem for backpropagation
sigmoid(-100) # it is close to zero but not exactly zero, so it is not a problem for backpropagation
```

```python
# always between -1 and 1, so it is not a problem for backpropagation
def tanh(x):
    return (math.exp(x) - math.exp(-x)) / (math.exp(x) + math.exp(-x))
```

```python
tanh(100) # it is close to one but not exactly one, so it is not a problem for backpropagation
tanh(0) # it is exactly 0, so it is not a problem for backpropagation
tanh(-100) # it is close to negative one but not exactly negative one, so it is not a problem for backpropagation
```

```python
def relu(x):
    return max(0,x)
```

```python
relu(-100) # it is exactly 0, so it is not a problem for backpropagation
relu(0) # it is exactly 0, so it is not a problem for back
relu(100) # it is exactly 100, so it is not a problem for backpropagation
```

```python
def leaky_relu(x):
    return max(0.01*x,x)
```

```python
leaky_relu(100) # it is exactly 0, so it is not a problem for backpropagation
leaky_relu(0) # it is exactly 0, so it is not a problem for back
leaky_relu(-12)
```
