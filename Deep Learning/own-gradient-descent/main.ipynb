```python
import numpy as np 
```

```python
X = np.array([
    [1,2],
    [2,5],
    [12,42],
    [124,121]
])

y = np.array([
    [1],
    [2],
    [5],
    [10]
])
```

```python
class GradientDescent:

    def __init__(self, X, y):
        self.X = X              # (n_samples, n_features)
        self.y = y              # (n_samples, 1)

        self.n_samples, self.n_features = X.shape
        self.weights = np.random.randn(self.n_features, 1)
        self.bias = 0

    def fit(self, learning_rate=0.01):

        # Forward
        y_pred = np.dot(self.X, self.weights) + self.bias

        # Error
        error = y_pred - self.y

        # Gradients
        dw = (2 / self.n_samples) * np.dot(self.X.T, error)
        db = (2 / self.n_samples) * np.sum(error)

        # Update
        self.weights -= learning_rate * dw
        self.bias -= learning_rate * db

        print("weights:", self.weights)
        print("dw:", dw)
```

```python
model = GradientDescent(X, y)

model.fit(learning_rate=0.5)
```
