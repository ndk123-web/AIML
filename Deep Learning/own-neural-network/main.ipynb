```python
import numpy as np
```

```python
class MyNeuralNetwork:

    def __init__(self):
        self.w1 = 1
        self.w2 = 1
        self.bias = 0

    def log_loss(y_true, y_predicted):
        epsilon = 1e-15
        y_predicted_new = [max(i,epsilon) for i in y_predicted]
        y_predicted_new = [min(i,1-epsilon) for i in y_predicted_new]
        y_predicted_new = np.array(y_predicted_new)
        return -np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))

    def sigmoid(X):
        return 1 / (1 + np.exp(-X))

    def gradient_descent(self, feature1 , feature2, y_truth, epochs, loss_threashold):
        w1 = 1
        w2 = 1
        bias = 0
        learning_rate = 0.05
        n = len(feature1)

        for i in range(epochs):
            weighted_sum = w1 * feature1 + w2 * feature2 + bias
            y_predicted = self.sigmoid(weighted_sum)
            loss = self.log_loss(y_truth,y_predicted)

            w1d = (1/n)*np.dot(np.transpose(feature1),(y_predicted-y_truth)) 
            w2d = (1/n)*np.dot(np.transpose(feature2),(y_predicted-y_truth)) 
            bias_d = np.mean(y_predicted-y_truth)

            w1 = w1 - learning_rate * w1d
            w2 = w2 - learning_rate * w2d
            bias = bias - learning_rate * bias_d

            if epochs > 1000 and i % 50 == 0:
                print(f"Loss: {loss}, w1: {w1}, w2: {w2}, bias: {bias}")

            if loss <= loss_threashold:
                break

        return (w1, w2, bias, loss)

    def predict(self, X):
        weighted_sum = self.w1 * X["age"] + self.w2 * X['affordibility'] + self.bias
        return self.sigmoid(weighted_sum)

    def fit(self, X, y):
        self.w1, self.w2, self.bias, self.loss = self.gradient_descent(X["age"], X["affordibility"],y,2000,0.46)
        print(f"Final weights and bias: w1: {self.w1}, w2: {self.w2}, bias: {self.bias}")
```
