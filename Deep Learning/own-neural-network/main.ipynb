#### Single Neuron From Scratch

- Each Neuron has weight (inputSize length)
- and 1 bias
- w1.x1 + w2.x2 + ... wn.xn + b


````python
input = [1.2, 2.4, 3.2, 1.2]

weight = [1, 3, 2, 5]
bias = 1

single_neuron_prediction_weighted_sum = (
    input[0] * weight[0]
    + input[1] * weight[1]
    + input[2] * weight[2]
    + input[3] * weight[3]
)

# To Add Non Linearity
signmoid_activation_function = 1 / (1 + np.exp(-single_neuron_prediction_weighted_sum))

print(
    f"Weighted Sum For Single Neuron: {single_neuron_prediction_weighted_sum}, Activation Function result: {signmoid_activation_function}"
)
````

#### Single Layer Of Neurons (No Numpy)

- as we said that each neuron has weights respectively according to the input
- and bias for each neuron
- we take input from previous layer's output and on that we are performing

`for below assume previously has 4 neurons will be giving result to the next layer's single neuron`

- we will gate the prbability of


````python
input = [1.2, 2.4, 3.2, 1.2]

weights = [
    [0.1, 0.4, 0.24, 1.2],
    [0.51, 0.41, 0.224, 1.25],
    [0.123, 0.234, 20.2234, 1.22],
    [0.51, 0.41, 0.2, 1.6],
]
biases = [3, 4, 5, 1]

layer_output = []

# Iterate over weights and biases
for neuron_weight, neuron_bias in zip(weights, biases):
    neuron_output_weighted_sum = 0

    # go over input and specific neuron weights
    # We Are indirectly doin the (DOT PRODUCT)
    for inp, neu_wei in zip(input, neuron_weight):
        # w1*x1 + ... w2*x2
        neuron_output_weighted_sum = neuron_output_weighted_sum + (inp * neu_wei)

    # add bias to that
    neuron_output_weighted_sum = neuron_output_weighted_sum + neuron_bias
    activation_function = 1 / (1 + np.exp(-neuron_output_weighted_sum))

    layer_output.append(activation_function)

layer_output
````

- as we said that each neuron has weights respectively according to the input

- we take input from previous layer's output and on that we are performing

`for below assume previously has 4 neurons will be giving result to the next layer's single neuron`

`Dot Product is important`

see
`case 1`

- assume weights has (3,4) and input has (4, 1) means 1 row with 4 columns (python treats 1 row matrix with (n, ) where n is columns)

```python
np.dot(weights, input)
```

- it will work because (3,4) \* (4,1) = (3,1) matrix

`case 2`

- assume now

```python
np.dot(input, weights)
```

- it will give error because (4,1) \* (3,4) not matched (error)


````python
input = np.array([[1.2, 2.4, 3.2, 1.2]])

weights = [
    [0.1, 0.4, 0.24, 1.2],
    [0.51, 0.41, 0.224, 1.25],
    [0.123, 0.234, 20.2234, 1.22],
]
biases = [3, 4, 5]

layer_output = []

# weighted sum (1st weights tells that we want 3*0 )
neuron_output_weighted_sum = np.dot(input, np.array(weights).T) + biases

# activation function (introduces non linearity (0-1))
activation_function = 1 / (1 + np.exp(-neuron_output_weighted_sum))

layer_output.append(activation_function)
layer_output
````

#### Object Layer   


````python
from numpy.char import lower
import numpy as np 

X = [[1, 2, 3, 4.5], [2, 3, 4, 1], [1, 1, 2, 4], [1, 5, 2, 6]]

np.random.seed(42)

class Layer_Dense:

    def __init__(self, n_inputs, n_neurons, activation_fn = "relu"):
        # each Column represents the weight size per neuron (for input) 
        self.weights = np.random.randn(n_inputs, n_neurons)
        self.biases = np.zeros((1, n_neurons))
        self.activation_function = activation_fn

    def forward(self, X):
        weighed_sum = np.dot(X ,self.weights) + self.biases

        if lower(self.activation_function) == "sigmoid":
            activation_function = self.sigmoid(weighed_sum)
        
        elif lower(self.activation_function) == "relu":
            activation_function = self.relu(weighed_sum)
 
        return activation_function
    
    def sigmoid(self, w):
        return 1 / (1 + np.exp(-w)) 
    
    def relu(self, w):
        return  np.maximum(0,w)  
    
    def softmax(self, w):
        pass 


# Input -> HidderLayer 1
layer_1 = Layer_Dense(4, 5, activation_fn = "relu")

# HiddenLayer 2 -> Outpue Neuron(2)
layer_2 = Layer_Dense(5, 2, activation_fn = "sigmoid")  

layer_1_output = layer_1.forward(X)
layer_2_output = layer_2.forward(layer_1_output)  

print(layer_1_output)
print()
print(layer_2_output)  
````

````python
np.random.seed(42)

def spiral_data(points, classes):
    X = np.zeros((points*classes, 2))
    y = np.zeros(points*classes, dtype='uint8')
    for class_number in range(classes):
        ix = range(points*class_number, points*(class_number+1))
        r = np.linspace(0.0, 1, points)  # radius
        t = np.linspace(class_number*4, (class_number+1)*4, points) + np.random.randn(points)*0.2
        X[ix] = np.c_[r*np.sin(t*2.5), r*np.cos(t*2.5)]
        y[ix] = class_number
    return X, y

class Dense_Layer:

    def __init__(self, n_inputs, n_neurons):
        self.weights = 0.10 * np.random.randn(n_inputs, n_neurons)
        self.bias = np.zeros((1, n_neurons))

    def forward(self, inputs):
        self.output = np.dot(inputs, self.weights) + self.bias

layer_1 = Dense_Layer(4, 5)
layer_1.forward(X)

layer_2 = Dense_Layer(5, 2)
layer_2.forward(layer_1.output)

print(layer_1.output)  # for each sample each neuron output
print()
print(layer_2.output)  # for each sample each neuron output
````

````python
# 3 classification with 100 points 
X, y = spiral_data(100, 3)
````

## Loss
1. For multi class we used categorical cross entropy loss function (required one hot encoding)
2. for binary class we used binary cross entropy
3. for regression or that we used (mse, mae)

````python
def categorical_cross_entropy(pred, act):
    
    # very important to prevent from log(0) because if its 0 then its inf and further output breks as Nan
    pred = np.clip(pred, 1e-7, 1 - 1e-7) 

    sum = np.sum(pred * act, axis=1)
    output = np.round(-np.mean((np.log(sum))), 4)

def binary_cross_entropy(pred, act):
    
    # very important to prevent from log(0) because if its 0 then its inf and further output breks as Nan
    pred = np.clip(pred, 1e-7, 1 - 1e-7) 

    output = -np.mean((act) * np.log(pred) + (1 - act) * np.log(1 - pred))
````
