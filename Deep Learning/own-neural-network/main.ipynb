#### Single Neuron From Scratch

- Each Neuron has weight (inputSize length)
- and 1 bias
- w1.x1 + w2.x2 + ... wn.xn + b


````python
input = [1.2, 2.4, 3.2, 1.2]

weight = [1, 3, 2, 5]
bias = 1

single_neuron_prediction_weighted_sum = (
    input[0] * weight[0]
    + input[1] * weight[1]
    + input[2] * weight[2]
    + input[3] * weight[3]
)

# To Add Non Linearity
signmoid_activation_function = 1 / (1 + np.exp(-single_neuron_prediction_weighted_sum))

print(
    f"Weighted Sum For Single Neuron: {single_neuron_prediction_weighted_sum}, Activation Function result: {signmoid_activation_function}"
)
````

#### Single Layer Of Neurons (No Numpy)

- as we said that each neuron has weights respectively according to the input
- and bias for each neuron
- we take input from previous layer's output and on that we are performing

`for below assume previously has 4 neurons will be giving result to the next layer's single neuron`

- we will gate the prbability of


````python
input = [1.2, 2.4, 3.2, 1.2]

weights = [
    [0.1, 0.4, 0.24, 1.2],
    [0.51, 0.41, 0.224, 1.25],
    [0.123, 0.234, 20.2234, 1.22],
    [0.51, 0.41, 0.2, 1.6],
]
biases = [3, 4, 5, 1]

layer_output = []

# Iterate over weights and biases
for neuron_weight, neuron_bias in zip(weights, biases):
    neuron_output_weighted_sum = 0

    # go over input and specific neuron weights
    # We Are indirectly doin the (DOT PRODUCT)
    for inp, neu_wei in zip(input, neuron_weight):
        # w1*x1 + ... w2*x2
        neuron_output_weighted_sum = neuron_output_weighted_sum + (inp * neu_wei)

    # add bias to that
    neuron_output_weighted_sum = neuron_output_weighted_sum + neuron_bias
    activation_function = 1 / (1 + np.exp(-neuron_output_weighted_sum))

    layer_output.append(activation_function)

layer_output
````

#### Single Layer Of Neurons (Numpy)

- as we said that each neuron has weights respectively according to the input
- and bias for each neuron
- we take input from previous layer's output and on that we are performing

`for below assume previously has 4 neurons will be giving result to the next layer's single neuron`

- we will gate the prbability of

`Dot Product is important`

see
`case 1`

- assume weights has (3,4) and input has (4, 1) means 1 row with 4 columns (python treats 1 row matrix with (n, ) where n is columns)

```python
np.dot(weights, input)
```

- it will work because (3,4) \* (4,1) = (3,1) matrix

`case 2`

- assume now

```python
np.dot(input, weights)
```

- it will give error because (4,1) \* (3,4) not matched (error)


````python
input = [1.2, 2.4, 3.2, 1.2]

weights = [
    [0.1, 0.4, 0.24, 1.2],
    [0.51, 0.41, 0.224, 1.25],
    [0.123, 0.234, 20.2234, 1.22],
]
biases = [3, 4, 5]

layer_output = []

# weighted sum (1st weights tells that we want 3*0 )
neuron_output_weighted_sum = np.dot(weights, input) + biases

# activation function (introduces non linearity (0-1))
activation_function = 1 / (1 + np.exp(-neuron_output_weighted_sum))

layer_output.append(activation_function)
layer_output
````

2. Multi Layers


````python
inputs = [[1, 2, 3, 4.5], [2, 3, 4, 1], [1, 1, 2, 4], [1, 5, 2, 6]]

layer_1_weights = [
    [0.1, 0.4, 0.24, 1.2],
    [0.51, 0.41, 0.224, 1.25],
    [0.123, 0.234, 20.2234, 1.22],
]
layer_2_weights = [
    [0.51, 0.41, 0.224],
    [0.123, 0.234, 20.2234],
    [0.1, 0.4, 0.24],
]
layer_1_biases = [3, 4, 5]
layer_2_biases = [1, 5, 3]

# Transpose the Weights and then all will be fine then
layer_1_weighted_sum = np.dot(inputs, np.array(layer_1_weights).T) + layer_1_biases

# always remember the input size for second layer will be previous neuron layer size 
layer_2_weighted_sum = (
    np.dot(np.array(layer_1_weighted_sum), np.array(layer_2_weights).T)
    + layer_2_biases
)

print(layer_1_weighted_sum)
print()
print(layer_2_weighted_sum)
````

#### Object Layer   


````python
import numpy as np 

X = [[1, 2, 3, 4.5], [2, 3, 4, 1], [1, 1, 2, 4], [1, 5, 2, 6]]

np.random.seed(42)

class Layer_Dense:

    def __init__(self, n_inputs, n_neurons):

        # each Column represents the weight size per neuron (for input) 
        self.weights = np.random.randn(n_inputs, n_neurons)
        self.biases = np.zeros((1, n_neurons))

    def forward(self, X):
        weighed_sum = np.dot(X ,self.weights)
        activation_function = 1 / (1 + np.exp(-weighed_sum))
        return activation_function


# Input -> HidderLayer 1
layer_1 = Layer_Dense(4, 5)

# HiddenLayer 2 -> Outpue Neuron(2)
layer_2 = Layer_Dense(5, 2)  

layer_1_output = layer_1.forward(X)
layer_2_output = layer_2.forward(layer_1_output)  

print(layer_1_output)
print()
print(layer_2_output)  
````
