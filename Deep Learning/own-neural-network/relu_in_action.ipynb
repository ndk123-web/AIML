```python
import time
from IPython.display import clear_output, display

def relu(x):
    return np.maximum(0, x)

x = np.linspace(-5, 5, 100)

fig, ax = plt.subplots(figsize=(6, 4))
w = 1
b = 1

# if we decrease the weight then relu
for i in range(1, 50):
    tw = w - i # 0, -1, -2, ... 
    tb = b + i  

    z = tw * x + b
    a = relu(z)

    ax.clear()
    ax.plot(x, z, '--', label="Z")
    ax.plot(x, a, label="ReLU")
    ax.set_title(f"w={w:.2f}")
    ax.legend()

    clear_output(wait=True)
    display(fig)
    time.sleep(1)

plt.close(fig)
```

```python
import time
from IPython.display import clear_output, display

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

x = np.linspace(-5, 5, 100)

fig, ax = plt.subplots(figsize=(6, 4))
w = 1
b = 1

# if we decrease the weight then relu
for i in range(1, 50):
    tw = w + i # 0, -1, -2, ... 
    tb = b - i  

    z = w * x + tb
    a = sigmoid(z)

    ax.clear()
    ax.plot(x, z, '--', label="Z")
    ax.plot(x, a, label="Sigmoid")
    ax.set_title(f"w={w:.2f}")
    ax.legend()

    clear_output(wait=True)
    display(fig)
    time.sleep(1)

plt.close(fig)
```
